## 4. Statistical Significance Testing Framework

Comprehensive statistical testing with effect sizes and multiple comparison corrections.

```python
# Initialize statistical analyzer
stat_analyzer = StatisticalAnalyzer(alpha=0.05)

# Create binary diabetes outcome for initial testing
df_processed['Diabetes_Binary'] = (df_processed['Diabetes_012'] == 2).astype(int)

# Test 1: Physical Activity vs Diabetes (Chi-square test)
phys_diabetes_result = stat_analyzer.chi_square_test(
    df_processed, 'PhysActivity', 'Diabetes_Binary'
)

print("Physical Activity vs Diabetes Association:")
print(f"Chi-square statistic: {phys_diabetes_result['chi2_statistic']:.4f}")
print(f"P-value: {phys_diabetes_result['p_value']:.2e}")
print(f"Effect size (Cramer's V): {phys_diabetes_result['effect_size_cramers_v']:.4f}")
print(f"Statistically significant: {phys_diabetes_result['significant']}")
print()
```

```python
# Test 2: BMI differences across diabetes groups (Kruskal-Wallis test)
bmi_diabetes_result = stat_analyzer.kruskal_wallis_test(
    df_processed, 'Diabetes_012', 'BMI'
)

print("BMI Differences Across Diabetes Groups:")
print(f"Kruskal-Wallis H-statistic: {bmi_diabetes_result['h_statistic']:.4f}")
print(f"P-value: {bmi_diabetes_result['p_value']:.2e}")
print(f"Effect size (eta-squared): {bmi_diabetes_result['effect_size_eta_squared']:.4f}")
print(f"Statistically significant: {bmi_diabetes_result['significant']}")
print()

# Test 3: Mental health by gender (Mann-Whitney U test)
mental_gender_result = stat_analyzer.mann_whitney_u_test(
    df_processed, 'Sex', 'MentHlth'
)

print("Mental Health Differences by Gender:")
print(f"Mann-Whitney U-statistic: {mental_gender_result['u_statistic']:.4f}")
print(f"P-value: {mental_gender_result['p_value']:.4f}")
print(f"Effect size (r): {mental_gender_result['effect_size_r']:.4f}")
print(f"Statistically significant: {mental_gender_result['significant']}")
```

```python
# Correlation analysis with significance testing
correlation_result = stat_analyzer.correlation_with_significance(
    df_processed, method='spearman'
)

print("Significant Correlations (Spearman):")
for corr in correlation_result['significant_correlations'][:10]:  # Top 10
    print(f"{corr['variable_1']} vs {corr['variable_2']}: "
          f"r = {corr['correlation']:.4f}, p = {corr['p_value']:.4f} "
          f"({corr['strength']})")
```

```python
# Apply multiple testing correction
correction_results = stat_analyzer.multiple_testing_correction(method='bonferroni')

print("Multiple Testing Correction Results:")
print(f"Correction method: {correction_results['correction_method']}")
print(f"Number of tests: {correction_results['n_tests']}")
print(f"Significant tests (original): {correction_results['n_significant_original']}")
print(f"Significant tests (corrected): {correction_results['n_significant_corrected']}")

# Generate comprehensive statistical report
statistical_report = stat_analyzer.generate_report()
print("\n" + "="*50)
print(statistical_report)
```

## 5. Multivariate Analysis with Confounding Control

Advanced epidemiological analysis controlling for confounding variables.

```python
# Initialize multivariate analyzer
multivariate_analyzer = MultivariateAnalyzer(df_processed)

# Adjusted analysis: Physical activity effect on diabetes, controlling for confounders
confounders = ['Age', 'BMI', 'Income', 'Sex']
adjusted_result = multivariate_analyzer.adjusted_analysis(
    outcome='Diabetes_Binary',
    exposure='PhysActivity', 
    confounders=confounders,
    method='logistic'
)

print("Adjusted Analysis: Physical Activity and Diabetes Risk")
print(f"Crude OR: {adjusted_result['crude_analysis']['odds_ratio']:.4f}")
print(f"Adjusted OR: {adjusted_result['adjusted_analysis']['odds_ratio']:.4f}")
print(f"Confounding assessment: {adjusted_result['confounding_assessment']['interpretation']}")
print(f"Multicollinearity detected: {adjusted_result['multicollinearity_check']['multicollinearity_detected']}")
```

```python
# Stratified analysis to examine effect modification
stratified_result = multivariate_analyzer.stratified_analysis(
    outcome='Diabetes_Binary',
    exposure='PhysActivity',
    stratifying_variable='Sex'
)

print("Stratified Analysis by Gender:")
print(f"Effect modification present: {stratified_result['effect_modification_present']}")

for stratum, results in stratified_result['strata_results'].items():
    gender = 'Female' if stratum == 0.0 else 'Male'
    or_value = results['odds_ratio']
    ci = results['confidence_interval']
    print(f"{gender}: OR = {or_value:.4f} (95% CI: {ci[0]:.4f}-{ci[1]:.4f})")
```

```python
# Interaction analysis
interaction_result = multivariate_analyzer.interaction_analysis(
    outcome='Diabetes_Binary',
    main_exposure='BMI',
    modifier='PhysActivity',
    method='logistic'
)

print("Interaction Analysis: BMI and Physical Activity")
print(f"Interaction coefficient: {interaction_result['interaction_coefficient']:.4f}")
print(f"Interaction p-value: {interaction_result['interaction_p_value']:.4f}")
print(f"Interpretation: {interaction_result['interpretation']}")
print(f"Likelihood ratio test significant: {interaction_result['likelihood_ratio_test']['significant']}")
```

## 6. Predictive Modeling Pipeline

Professional machine learning pipeline with clinical validation metrics.

```python
# Initialize predictive modeling pipeline
ml_pipeline = PredictiveModelPipeline(target_column='Diabetes_012', random_state=42)

# Prepare data for modeling
model_data = ml_pipeline.prepare_data(df_processed, test_size=0.2)

print("Model Data Preparation:")
print(f"Training samples: {model_data['n_train_samples']:,}")
print(f"Test samples: {model_data['n_test_samples']:,}")
print(f"Features: {model_data['n_features']}")
print(f"Class distribution: {model_data['class_distribution']}")
```

```python
# Initialize and optimize multiple models
models_config = ml_pipeline.initialize_models()

print("Training and optimizing models...")
print("Models to train:", list(models_config.keys()))

# Train and optimize models
optimization_results = ml_pipeline.train_and_optimize_models(
    model_data['X_train'], 
    model_data['y_train'], 
    cv_folds=5
)

print("\nModel Optimization Results:")
for model_name, results in optimization_results.items():
    print(f"{model_name}: CV ROC-AUC = {results['best_cv_score']:.4f}")
```

```python
# Evaluate models on test set
evaluation_results = ml_pipeline.evaluate_models(
    model_data['X_test'], 
    model_data['y_test']
)

print("Test Set Performance:")
print("Model".ljust(20), "Accuracy".ljust(10), "Precision".ljust(10), 
      "Recall".ljust(10), "F1-Score".ljust(10), "ROC-AUC".ljust(10))
print("-" * 70)

for model_name, results in evaluation_results.items():
    metrics = results['metrics']
    print(f"{model_name}".ljust(20),
          f"{metrics['accuracy']:.4f}".ljust(10),
          f"{metrics['precision']:.4f}".ljust(10),
          f"{metrics['recall']:.4f}".ljust(10),
          f"{metrics['f1_score']:.4f}".ljust(10),
          f"{metrics['roc_auc']:.4f}".ljust(10))
```

```python
# Display clinical performance metrics
print("\nClinical Performance Metrics:")
print("=" * 50)

for model_name, results in evaluation_results.items():
    clinical = results['clinical_metrics']
    print(f"\n{model_name.upper()}:")
    print(f"  Sensitivity (Recall): {clinical['sensitivity']:.4f}")
    print(f"  Specificity: {clinical['specificity']:.4f}")
    print(f"  Positive Predictive Value: {clinical['positive_predictive_value']:.4f}")
    print(f"  Negative Predictive Value: {clinical['negative_predictive_value']:.4f}")
    print(f"  Number Needed to Screen: {clinical['number_needed_to_screen']:.1f}")
```

```python
# Select best model and analyze feature importance
best_model_name, best_model = ml_pipeline.select_best_model()
feature_importance = ml_pipeline.feature_importance_analysis()

print(f"Best Model: {best_model_name}")
print(f"Test ROC-AUC: {evaluation_results[best_model_name]['metrics']['roc_auc']:.4f}")

if best_model_name in feature_importance:
    print("\nTop 10 Most Important Features:")
    importance_data = feature_importance[best_model_name]
    
    if importance_data['type'] == 'feature_importances':
        for i, feature in enumerate(importance_data['importances'][:10]):
            print(f"{i+1:2d}. {feature['feature']}: {feature['importance']:.4f}")
    elif importance_data['type'] == 'coefficients':
        for i, feature in enumerate(importance_data['coefficients'][:10]):
            print(f"{i+1:2d}. {feature['feature']}: {feature['coefficient']:.4f}")
```

```python
# Create comprehensive model visualizations
ml_pipeline.create_model_visualizations(
    model_data['X_test'], 
    model_data['y_test'], 
    save_plots=False
)
```

## 7. Advanced Model Validation and Performance Assessment

Clinical-grade model validation with comprehensive performance metrics.

```python
# Initialize advanced analytics engine
validator = AdvancedAnalyticsEngine(random_state=42)

# Perform comprehensive model validation
validation_results = validator.comprehensive_model_validation(
    best_model,
    model_data['X_train'],
    model_data['y_train'],
    cv_folds=10,
    validation_type='stratified'
)

print("Comprehensive Model Validation Complete")
print(f"Cross-validation strategy: {validation_results['validation_metadata']['cv_strategy']}")
print(f"Number of folds: {validation_results['validation_metadata']['n_folds']}")
```

```python
# Display cross-validation performance summary
cv_results = validation_results['cross_validation_results']

print("Cross-Validation Performance (10-Fold):")
print("Metric".ljust(15), "Mean".ljust(10), "Std".ljust(10), "95% CI".ljust(20))
print("-" * 55)

key_metrics = ['test_accuracy', 'test_precision', 'test_recall', 'test_f1', 'test_roc_auc']
for metric in key_metrics:
    if metric in cv_results:
        result = cv_results[metric]
        ci = result['confidence_interval_95']
        metric_name = metric.replace('test_', '').replace('_', ' ').title()
        print(f"{metric_name}".ljust(15), 
              f"{result['mean']:.4f}".ljust(10),
              f"{result['std']:.4f}".ljust(10),
              f"[{ci[0]:.4f}, {ci[1]:.4f}]".ljust(20))
```

```python
# Display clinical validation metrics
clinical_metrics = validation_results['clinical_metrics']

print("\nClinical Validation Metrics (Cross-Validation):")
print("Metric".ljust(25), "Mean".ljust(10), "Std".ljust(10))
print("-" * 45)

clinical_names = {
    'sensitivity': 'Sensitivity (Recall)',
    'specificity': 'Specificity',
    'positive_predictive_value': 'Positive Predictive Value',
    'negative_predictive_value': 'Negative Predictive Value',
    'number_needed_to_treat': 'Number Needed to Treat'
}

for metric_key, metric_name in clinical_names.items():
    if metric_key in clinical_metrics:
        result = clinical_metrics[metric_key]
        print(f"{metric_name}".ljust(25),
              f"{result['mean']:.4f}".ljust(10),
              f"{result['std']:.4f}".ljust(10))
```

```python
# Display learning curve analysis
learning_curves = validation_results['learning_curves']

print("\nLearning Curve Analysis:")
print(f"Interpretation: {learning_curves['interpretation']}")

# Plot learning curves
plt.figure(figsize=(10, 6))
train_sizes = learning_curves['train_sizes']
train_scores_mean = learning_curves['train_scores_mean']
train_scores_std = learning_curves['train_scores_std']
test_scores_mean = learning_curves['test_scores_mean']
test_scores_std = learning_curves['test_scores_std']

plt.plot(train_sizes, train_scores_mean, 'o-', color='blue', label='Training Score')
plt.fill_between(train_sizes, 
                 np.array(train_scores_mean) - np.array(train_scores_std),
                 np.array(train_scores_mean) + np.array(train_scores_std),
                 alpha=0.3, color='blue')

plt.plot(train_sizes, test_scores_mean, 'o-', color='red', label='Cross-Validation Score')
plt.fill_between(train_sizes,
                 np.array(test_scores_mean) - np.array(test_scores_std),
                 np.array(test_scores_mean) + np.array(test_scores_std),
                 alpha=0.3, color='red')

plt.xlabel('Training Set Size')
plt.ylabel('ROC-AUC Score')
plt.title('Learning Curves')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Feature importance stability analysis
feature_stability = validation_results['feature_importance_stability']

if 'overall_stability' in feature_stability:
    print("\nFeature Importance Stability Analysis:")
    print(f"Overall stability score: {feature_stability['overall_stability']:.4f}")
    print(f"Interpretation: {feature_stability['interpretation']}")
    
    if feature_stability['unstable_features']:
        print(f"Unstable features: {', '.join(feature_stability['unstable_features'][:5])}")
    
    # Plot top 10 most stable features
    if 'feature_stability_ranking' in feature_stability:
        stability_df = pd.DataFrame(feature_stability['feature_stability_ranking'])
        top_10_stable = stability_df.head(10)
        
        plt.figure(figsize=(10, 6))
        plt.barh(range(len(top_10_stable)), top_10_stable['stability_score'])
        plt.yticks(range(len(top_10_stable)), top_10_stable['feature'])
        plt.xlabel('Stability Score')
        plt.title('Top 10 Most Stable Features')
        plt.gca().invert_yaxis()
        plt.grid(True, alpha=0.3)
        plt.show()
```

```python
# Bootstrap validation for robust confidence intervals
print("Performing bootstrap validation...")
bootstrap_results = validator.bootstrap_validation(
    best_model,
    model_data['X_train'],
    model_data['y_train'],
    n_bootstrap=100  # Reduced for demo purposes
)

print("\nBootstrap Validation Results:")
print("Metric".ljust(15), "Mean".ljust(10), "Std".ljust(10), "95% CI".ljust(20))
print("-" * 55)

for metric, results in bootstrap_results['bootstrap_results'].items():
    ci_lower = results['ci_2.5']
    ci_upper = results['ci_97.5']
    print(f"{metric.title()}".ljust(15),
          f"{results['mean']:.4f}".ljust(10),
          f"{results['std']:.4f}".ljust(10),
          f"[{ci_lower:.4f}, {ci_upper:.4f}]".ljust(20))
```
